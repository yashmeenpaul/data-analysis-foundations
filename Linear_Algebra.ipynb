```python
#2 to create 4X4 matrix using numpy with random numbers in the range of 1-100 and 
#print all the odd numbers using looping
#print all the even numbers
```

```python
a=np.random.randint(1,100,(4,4))
a
```

```python
print("\nOdd Numbers:")
for row in a:
    for num in row:
        if num % 2 !=0:
            print(num, end=" ")
```

```python
print("\nEven Numbers:")
for row in a:
    for num in row:
        if num % 2 ==0:
            print(num, end=" ")
```

```python
#Create a NumPy array of 15 random integers between 1 and 100.
#Using a while loop, find and print the maximum value in the array.without using any inbuilt function
```

```python
import numpy as np

arr = np.random.randint(1, 101, 15)
print("Array:", arr)
```

```python
max_val = arr[0] #90
i = 1 

while i < len(arr):
    if arr[i] > max_val: #80>90  
        max_val = arr[i]   
    i += 1 #i=i+1

print("Maximum value:", max_val)
```
